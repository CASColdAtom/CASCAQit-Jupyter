# JupyterLab 原生 GHZ 态可视化

这个离线示例直接向 Notebook 输出 CASCAQit JSON MIME 数据，由 JupyterLab 扩展在当前页面渲染量子线路、执行结果和计数直方图。

```python
from cascaqit import Circuit, build_counts_histogram

from cascaqit_jupyter import (
    display_program,
    display_result,
    display_visualization,
)

SHOTS = 256
SEED = 2026
```

```python
circuit = Circuit(
    ["q0", "q1", "q2"],
    program_id="program.demo.ghz",
)
circuit.h("q0")
circuit.cx("q0", "q1")
circuit.cx("q1", "q2")
circuit.measure_all(key="m")

program = circuit.to_program()
result = circuit.run(
    shots=SHOTS,
    seed=SEED,
    return_probabilities=True,
)
histogram = build_counts_histogram(result)

assert sum(result.counts.values()) == SHOTS
assert set(result.counts).issubset({"000", "111"})
```

## 量子线路

```python
display_program(program)
```

## 本地执行结果

```python
display_result(result)
```

## 测量计数直方图

```python
display_visualization(histogram)
```
